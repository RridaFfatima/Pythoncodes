## Scenario 01: University Course Registration System

A university wants to manage course registrations for students. Each student can register for multiple courses, but duplicate registrations must not be allowed.

**Your task:**
- Ask for the student's name and the courses they want to register for.
- Store the registered courses in a **list**.
- Use a **set** to identify and prevent duplicate course registrations.
- Store each course's code and credit hours in a **tuple**.
- Use a **dictionary** to maintain course details, including course codes and credit hours.
- Display the final registered courses, total credit hours, and any duplicate entries rejected.

```python
course_details={"PF":("h03l",3),"ICT":("a014",2),"PST":("h506", 2)}
name=input("Enter your name: ")
noofcourses=int(input("enter the number of courses you want to enter: "))
registered_courses=[]
registered_set=set()
duplicates =[]
for i in range(noofcourses):
    course_name=input("Ente the course name:(PF,ICT,PST) ").upper()
    if(course_name not in course_details):
        print("invaid course")
    elif( course_name in registered_set):
        print("course already presnet")
        duplicates.append(course_name)
    else:
        registered_courses.append(course_name)
        registered_set.add(course_name)

print("student name: ",name)
print("registered courses: ",noofcourses)
print("details: ")
total_credit_hours = 0

for course in registered_courses:
    course_code, credits = course_details[course]
    print(course, "-", course_code, "-", credits, "credit hours")
    total_credit_hours += credits


print("total credit hours: "+ str(total_credit_hours) +" credit hours")    
print("duplicate courses: ",duplicates)
```

## Scenario 02: Inventory and Stock Management System

A small shop needs a program to monitor its products, prices, and available stock.

**Your task:**
- Allow the shopkeeper to enter product names, prices, and quantities.
- Use a **dictionary** to store product information, with product names as keys.
- Store each product's price and quantity together in a **tuple**.
- Maintain a **set** of unique product categories.
- Use a **list** to record products that need restocking when their quantity falls below five.
- Display the complete inventory, total inventory value, unique categories, and restocking list.
- Allow the shopkeeper to update stock quantities and prices.

```python

product_info = {}
categories = set()
restock = []

n = int(input("Enter how many products to enter: "))

for i in range(n):
    name = input("enter the product name: ").lower()
    price = float(input("enter the product price: "))
    quantity = int(input("enter the product quantity: "))
    category = input("enter the product category: ").lower()

    product_details = (price, quantity)
    product_info[name] = product_details
    categories.add(category)

    if quantity < 5:
        restock.append(name)

print("Complete Inventory:")
total_value = 0

for name in product_info:
    price, quantity = product_info[name]
    value = price * quantity
    total_value += value
    print(name, " Price:", price, " Quantity:", quantity)

print("Total inventory value:", total_value)
print("Unique categories:", categories)
print("Products needing restock:", restock)

updatename = input("Enter the name of product to update: ").lower()

if updatename in product_info:
    updatequantity = int(input("Enter the new quantity: "))
    update_price = float(input("Enter the updated price: "))

    product_info[updatename] = (update_price, updatequantity)

    
    if updatequantity < 5:
        if updatename not in restock:
            restock.append(updatename)
    else:
        if updatename in restock:
            restock.remove(updatename)
else:
    print("Product not found!")
total_value = 0

for name in product_info:
    price, quantity = product_info[name]
    total_value += price * quantity

print("Updated total inventory value:", total_value)
print("Updated Inventory:")
for name in product_info:
    price, quantity = product_info[name]
    print(name, " Price: ", price, " Quantity: ", quantity)

print("Updated restock list:", restock)
```

## Scenario 03: Event Registration and Attendance Management

A university society organizes an event and needs to manage registrations and attendance.

**Your task:**
- Accept participant names and their registration numbers.
- Use a **dictionary** to associate each registration number with a participant's name.
- Maintain a **set** of registered participants to prevent duplicate registrations.
- Store event details, such as event name, date, and venue, in a **tuple**.
- Use a **list** to record participants who attend the event.
- Display registered participants, attendees, absentees, and total attendance percentage.
- Ensure that unregistered participants cannot be marked as present.

```python

event_details = ("AWS community day", "22 November", "IEEE hall")

participants = {}
registered_set = set()
attendees = []

n = int(input("Enter the number of participants: "))

for i in range(n):
    reg_no = input("Enter registration number: ")
    name = input("Enter participant name: ")

    if reg_no in registered_set:
        print("Participant already registered")
    else:
        participants[reg_no] = name
        registered_set.add(reg_no)

print("Event Details:")
print("Event Name:", event_details[0])
print("Date:", event_details[1])
print("Venue:", event_details[2])

print("Registered Participants:")
for reg_no in participants:
    print(reg_no, " : ", participants[reg_no])

m = int(input("Enter number of participants to mark present: "))

for i in range(m):
    reg_no = input("Enter registration number: ")

    if reg_no not in registered_set:
        print("Unregistered participant!")
    elif reg_no in attendees:
        print("Attendance already marked")
    else:
        attendees.append(reg_no)
        print("Attendance marked")

print("Attendees:")
for reg_no in attendees:
    print(reg_no, ":", participants[reg_no])

print("Absentees:")
for reg_no in participants:
    if reg_no not in attendees:
        print(reg_no, ":", participants[reg_no])

total_registered = len(participants)
total_attendees = len(attendees)

if total_registered > 0:
    percentage = (total_attendees / total_registered) * 100
else:
    percentage = 0

print("Total Registered:", total_registered)
print("Total Attendance:", total_attendees)
print("Attendance Percentage:", percentage, "%")
```

## Scenario 04: Online Examination Result Analyzer

A teacher wants to analyze examination results for a class of students across multiple subjects.

**Your task:**
- Accept students' names and their marks in multiple subjects.
- Use a **dictionary** to store each student's subject-wise marks.
- Use a **list** to maintain the class roster.
- Use a **tuple** to store the names of subjects included in the examination.
- Use a **set** to identify the distinct grades achieved by the class, using these grade boundaries: A (80–100), B (70–79), C (60–69), D (50–59), and F (below 50).
- Calculate each student's total marks, average marks, and grade.
- Display the class topper, students who failed in at least one subject, and the class average.
- Validate that marks are between 0 and 100.

```python

subjects = ("COAL", "LA", "IDS")

students = {}
class_roster = []
grades = set()

n = int(input("Enter number of students: "))

for i in range(n):
    name = input("Enter student name: ")

    if name in class_roster:
        print("Student already entered")
    else:
        marks = {}
        for subject in subjects:
            mark = int(input("Enter marks in " + subject + ": "))

            while mark < 0 or mark > 100:
                mark=int(input("Invalid marks! Enter marks between 0 and 100"))
            marks[subject] = mark
        
        students[name] = marks
        class_roster.append(name)

total_class_marks = 0
total_class_students = 0
topper = ""
highest_average = -1
failed_students = []

print("Exam Results:")

for name in class_roster:
    total = 0
    failed = False

    for subject in subjects:
        mark = students[name][subject]
        total += mark

        if mark < 50:
            failed = True

    averagemarks = total / len(subjects)

    if averagemarks >= 80:
        grade = "A"
    elif averagemarks >= 70:
        grade = "B"
    elif averagemarks >= 60:
        grade = "C"
    elif averagemarks >= 50:
        grade = "D"
    else:
        grade = "F"

    grades.add(grade)
    total_class_marks += averagemarks
    total_class_students += 1

    if averagemarks > highest_average:
        highest_average = averagemarks
        topper = name

    if failed:
        failed_students.append(name)

    print("Name:", name)
    print("Subject-wise Marks:", students[name])
    print("Total Marks:", total)
    print("Average Marks:", averagemarks)
    print("Grade:", grade)

print("Class Topper:", topper)
print("Topper Average:", highest_average)
print("Students who failed in at least one subject:", failed_students)
print("Grades:", grades)

if total_class_students > 0:
    class_average = total_class_marks / total_class_students
else:
    class_average = 0

print("Class Average:", class_average)
```

## Scenario 05: Personal Expense Tracker and Spending Analysis

A student wants to track daily expenses and identify where their money is being spent.

**Your task:**
- Accept expense descriptions, categories, and amounts over multiple entries.
- Use a **list** to maintain the expense records.
- Use a **tuple** to represent each expense record, containing its description, category, and amount.
- Use a **dictionary** to calculate total expenditure for each category.
- Use a **set** to display all unique expense categories.
- Allow the student to search for expenses by category.
- Display total expenditure, category-wise spending, the highest individual expense, and the category with the highest total spending.
- Reject negative expense amounts and handle cases where no expenses have been recorded.

```python

expenses = []
categories = set()
category_totals = {}

n = int(input("Enter the number of expenses: "))

for i in range(n):
    description = input("enter expense description: ")
    category = input("enter expense category: ").lower()
    amount = float(input("enter expense amount: "))

    if amount < 0:
        print("Invalid amount, Negative expenses are not allowed")
    else:
        expense = (description, category, amount)
        expenses.append(expense)
        categories.add(category)

        if category in category_totals:
            category_totals[category] += amount
        else:
            category_totals[category] = amount

if len(expenses) == 0:
    print("No expenses have been recorded")

else:
    total_expenditure = 0
    highest_expense = expenses[0]
    highest_category = ""
    highest_category_total = -1

    for expense in expenses:
        description, category, amount = expense
        total_expenditure += amount

        if amount > highest_expense[2]:
            highest_expense = expense

    for category in category_totals:
        if category_totals[category] > highest_category_total:
            highest_category_total = category_totals[category]
            highest_category = category

    print("All expenses:")
    for expense in expenses:
        print(expense)

    print("Total expenditure:", total_expenditure)

    print("Categorywise spending:")
    for category in category_totals:
        print(category, ":", category_totals[category])

    print("Unique categories:", categories)

    print("Highest individual expense:")
    print("description:", highest_expense[0])
    print("category:", highest_expense[1])
    print("amount:", highest_expense[2])

    print("category with highest amount spent: ")
    print(highest_category, ":", highest_category_total)

    search_category = input("Enter category to search for expenses: ").lower()

    found = False


    for expense in expenses:
        if expense[1] == search_category:
            print("expenses in", search_category, ":",expense)
            found = True

    if not found:   
        print("no expenses found in this category")
```
